```python
# import os

with open('data.csv','w') as f:
    f.write('x / y\n')
    for val in range(101):
        f.write(f'{val}, {val**2}\n')

# os.startfile('data.csv')
```

```python
x_values = []
y_values = []

with open('data.csv', 'r') as f:
    next(f)     #skip the header
    for line in f:
        parts = line.split(', ')
        x_values.append(int(parts[0]))
        y_values.append(int(parts[1]))
```

```python
x_values
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots()
ax.plot(x_values, y_values, color='green', linestyle='--')
plt.show()
```
